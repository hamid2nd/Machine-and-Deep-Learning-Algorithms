```python
# Linear layer with sigmoid
# (example from lecture in code)

def sigmoid(x):
  return 1 / (1 + math.exp(-x))

# Forward pass
def forward(x0, w0, x1, w1, w2):
    s0 = w0 * x0
    s1 = w1 * x1
    s2 = s0 + s1
    s3 = s2 + w2
    L = sigmoid(s3)
    return L

# Backward pass
def backward(x0, w0, x1, w1, w2, L):
    dL = 1.0
    ds3 = dL * L * (1 - L)
    dw2 = ds3
    ds2 = ds3
    ds1 = ds2
    ds0 = ds2
    dw1 = ds1 * x1
    dx1 = ds1 * w1
    dw0 = ds0 * x0
    dx0 = ds0 * w0
    return  dx0, dw0, dx1, dw1, dw2


# Verify with example from the lecture
x0 = -1.0
w0 = 2.0
x1 = -2.0
w1 = -3.0
w2 = -3.0

L = forward(x0, w0, x1, w1, w2)
print("Output:", L)

dx0, dw0, dx1, dw1, dw2 = backward(x0, w0, x1, w1, w2, L)

print("Gradients:")
print("dx0:", dx0)
print("dw0:", dw0)
print("dx1:", dx1)
print("dw1:", dw1)
print("dw2:", dw2)
```
